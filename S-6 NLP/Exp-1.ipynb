```python
# Import tokenizers
from nltk.tokenize import (
    word_tokenize, 
    sent_tokenize, 
    regexp_tokenize,
    TweetTokenizer
)
```

```python
# Sample text for demonstration
sample_text = """NLP LAB 1 EXP 1 apply different tokenization techniques using nltk. 
Will learn about natural language tool-kit. This is a complex sentence, 
with punctuation! How about multiple sentences?"""

print("Original Text:")
print(sample_text)
```

```python
# Perform word tokenization
words = word_tokenize(sample_text)
print("\nWord Tokenization:")
print(words)
# This will split text into words, including punctuation
# Output example: ['NLP', 'LAB', '1', 'EXP', '1', 'apply', ...]
```

```python
# Perform sentence tokenization
sentences = sent_tokenize(sample_text)
print("\nSentence Tokenization:")
print(sentences)
# This will split text into sentences
# Output example: ['NLP LAB 1 EXP 1 apply...', 'Will learn about...', ...]
```

```python
# RegExp Tokenization - Words only
regexp_tokens = regexp_tokenize(sample_text, pattern=r'\b\w+\b')
print("\nRegExp Tokenization (Words Only):")
print(regexp_tokens)

# RegExp Tokenization - Words and Numbers
regexp_tokens_with_digits = regexp_tokenize(sample_text, pattern=r'\b\w+\b|\d+')
print("\nRegExp Tokenization (Words and Numbers):")
print(regexp_tokens_with_digits)
```

```python
# Custom tokenization (splitting by hyphens)
text_with_hyphens = "This is a hyphen-separated-text example"
custom_tokens = text_with_hyphens.split('-')
print("\nCustom Tokenization (Splitting by Hyphens):")
print(custom_tokens)
```

```python
# Tweet tokenization
tweet_text = "Check out this #awesome post by @user https://example.com!"
tweet_tokenizer = TweetTokenizer()
tweet_tokens = tweet_tokenizer.tokenize(tweet_text)
print("\nTweet Tokenization:")
print(tweet_tokens)
```

```python
# Compare different tokenization results
print("\nComparison of Different Tokenizers:")
print(f"Number of words (word_tokenize): {len(words)}")
print(f"Number of sentences: {len(sentences)}")
print(f"Number of regexp tokens (words only): {len(regexp_tokens)}")
```

```python
complex_text = """
Mr. Smith paid $30.00 for the book! 
Email: john@example.com
Website: https://www.example.com
#NLP is @awesome
This is a hyphen-separated-word.
"""

print("\nComplex Text Analysis:")
print("Word Tokenization:", word_tokenize(complex_text))
print("Sentence Tokenization:", sent_tokenize(complex_text))
print("Tweet Tokenization:", tweet_tokenizer.tokenize(complex_text))
```
