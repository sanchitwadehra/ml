```python
# Import required libraries
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

print("UID: 22BAI71060")

# 1. Data Preparation
print("\n=== Data Preparation ===")

# Load the MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Print initial data shapes
print("Initial shapes:")
print("Training data shape:", x_train.shape)
print("Training labels shape:", y_train.shape)
print("Testing data shape:", x_test.shape)
print("Testing labels shape:", y_test.shape)

# Normalize the images to range [0, 1]
x_train = x_train.astype('float32') / 255
x_test = x_test.astype('float32') / 255

# Flatten the images (reshape to 1D)
x_train = x_train.reshape((x_train.shape[0], 28 * 28))
x_test = x_test.reshape((x_test.shape[0], 28 * 28))

# One-hot encode the labels
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)

print("\nAfter preprocessing:")
print("Training data shape:", x_train.shape)
print("Training labels shape:", y_train.shape)

# 2. Model Definition
print("\n=== Model Creation ===")

# Create the sequential model
model = Sequential([
    Dense(512, activation='relu', input_shape=(28 * 28,)),
    Dense(512, activation='relu'),
    Dense(256, activation='relu'),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

# Display model summary
model.summary()

# 3. Model Compilation
print("\n=== Model Compilation ===")
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# 4. Model Training
print("\n=== Model Training ===")
history = model.fit(x_train, y_train,
                    epochs=10,
                    batch_size=128,
                    validation_split=0.2,
                    verbose=1)

# 5. Model Evaluation
print("\n=== Model Evaluation ===")
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f'Test accuracy: {test_accuracy:.4f}')
print(f'Test loss: {test_loss:.4f}')

# 6. Making Predictions
print("\n=== Making Predictions ===")
predictions = model.predict(x_test)

# Print results for first 5 test images
print("\nPredictions for first 5 images:")
for i in range(5):
    print(f'Image {i+1}:')
    print(f'Predicted digit: {np.argmax(predictions[i])}')
    print(f'Actual digit: {np.argmax(y_test[i])}')
    print()

# Print training history
print("\n=== Training History ===")
print("Final training accuracy:", history.history['accuracy'][-1])
print("Final validation accuracy:", history.history['val_accuracy'][-1])
```

```python


from sklearn.metrics import confusion_matrix  # Add this line

# Print confusion matrix
print("\n=== Confusion Matrix ===")
y_pred = np.argmax(predictions, axis=1)
y_true = np.argmax(y_test, axis=1)
conf_matrix = confusion_matrix(y_true, y_pred)
print("Confusion Matrix:")
print(conf_matrix)

# Print normalized confusion matrix
print("\nNormalized Confusion Matrix:")
conf_matrix_norm = conf_matrix.astype('float') / conf_matrix.sum(axis=1)[:, np.newaxis]
print(np.round(conf_matrix_norm, 2))
```

```python
# After your confusion matrix code, add:

def calculate_metrics(conf_matrix, class_idx):
    # For binary metrics, we treat the specified class as positive and all others as negative
    TP = conf_matrix[class_idx, class_idx]
    FP = np.sum(conf_matrix[:, class_idx]) - TP
    FN = np.sum(conf_matrix[class_idx, :]) - TP
    TN = np.sum(conf_matrix) - TP - FP - FN
    
    # Calculate precision, recall, and F1 score
    precision = TP / (TP + FP) if (TP + FP) != 0 else 0
    recall = TP / (TP + FN) if (TP + FN) != 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) != 0 else 0
    
    return {
        'TP': TP,
        'TN': TN,
        'FP': FP,
        'FN': FN,
        'Precision': precision,
        'Recall': recall,
        'F1': f1
    }

print("\n=== Detailed Metrics for Each Digit ===")
for digit in range(10):
    metrics = calculate_metrics(conf_matrix, digit)
    print(f"\nDigit {digit}:")
    print(f"True Positives (TP): {metrics['TP']}")
    print(f"True Negatives (TN): {metrics['TN']}")
    print(f"False Positives (FP): {metrics['FP']}")
    print(f"False Negatives (FN): {metrics['FN']}")
    print(f"Precision: {metrics['Precision']:.4f}")
    print(f"Recall: {metrics['Recall']:.4f}")
    print(f"F1 Score: {metrics['F1']:.4f}")
```
